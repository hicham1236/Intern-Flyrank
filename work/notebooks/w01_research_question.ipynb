## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

* Unit of analysis: A single URL / web page.

* The decision: Which specific pages should the content team prioritize for a rewrite or update this month?

* The action: Assigning a writer or editor to refresh the page content.

Cost of a wrong call:

* False Positive: Wasting expensive editorial hours rewriting a page that was doing fine or wouldn't benefit from an update.

* False Negative: Missing a decaying high-value page, leading to a permanent loss in organic traffic and revenue.

Why ML helps: A hand-written rule (like "update pages older than 6 months") is too rigid and creates a massive, unmanageable queue. ML can weigh multiple signals (position, age, historical CTR) to surface a prioritized, high-probability shortlist.

```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Create the label
df["is_declining"] = df["trend_direction"].str.lower().eq("down").astype(int)

# Number 1: The baseline rate
decline_rate = df["is_declining"].mean() * 100

# Number 2 & 3: Comparing average age
avg_age_declining = df[df["is_declining"] == 1]["content_age_days"].mean()
avg_age_stable = df[df["is_declining"] == 0]["content_age_days"].mean()

print(f"1. Baseline Rate: {decline_rate:.1f}% of the pages in this dataset are currently in a downward trend.")
print(f"2. Age Signal: Pages that are declining are on average {avg_age_declining:.0f} days old.")
print(f"3. Contrast: Stable/Growing pages are significantly younger, averaging {avg_age_stable:.0f} days old.")
```
